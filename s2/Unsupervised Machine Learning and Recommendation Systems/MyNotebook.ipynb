```python
# create a user item matrix with ratings of movies by users
ratings = pd.DataFrame({
    "harry potter": [5,5,1,4],
    "titanic":[1,2,5,2],
    "InterruptedError":[4,5,1,4],
    "narnia":[None,4,2,5]
}) 

index = ["user1", "user2", "user3", "user4"]
ratings.index = index
ratings
```

```python
# Mean center the ratings by subtracting the mean rating for each user from their ratings
ratings_centered = ratings.sub(ratings.mean(axis=1), axis=0)
ratings_centered
```

```python
cent = ratings.apply(lambda x: x - x.mean(), axis=1)
```

```python
cent
```

```python
ratings.apply(lambda x: print(x.mean()))
```

```python
cent # this is the mean centered ratings. we can then use this to calculate similarity between users and make recommendations based on that.
```

```python

# i dont want narnia to make predection
pred = cent.loc[:,["harry potter", "titanic", "InterruptedError"]]
pred
```

```python
# calculating cosine similarity between users based on the mean centered ratings and making recommendations based on that.
similarity = pred.dot(pred.T)
similarity
```

```python
import numpy as np
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
sim = cosine_similarity(pred)

```

```python
sim
```

```python
# Step 1: Your starting point (raw dot product)
# (Assuming pred is already filled with zeros via .fillna(0))
similarity = pred.dot(pred.T)

# Step 2: Calculate the magnitude (norm) for each user vector
# This gives you a single value per user representing their "vector length"
magnitudes = np.linalg.norm(pred, axis=1)

# Step 3: Normalize the similarity matrix
# np.outer(magnitudes, magnitudes) creates a grid where position (i, j) 
# is equal to (magnitude of user_i * magnitude of user_j)
cosine_sim_matrix = similarity / np.outer(magnitudes, magnitudes)
```

```python
cosine_sim_matrix # this is basically cos theeta in the dot product used for vectors
```

```python
sim = cosine_similarity(pred)
sim = pd.DataFrame(sim, index=pred.index, columns=pred.index)
k2 = sim["user1"].sort_values(ascending=False)
k2
```

```python
sum_weighted_ratings = (k2.loc["user2"] * cent.loc["user2", "narnia"]) + (k2.loc["user4"] * cent.loc["user4", "narnia"])
```

```python
# this is the predicted rating for user1 for narnia based on the ratings of user2 and user4 and their similarity to user1.
sum_weighted_ratings/(k2.loc["user2"] + k2.loc["user4"])
```

# Association Mining
1. Before calculating the rules, you need the Support values (how often things appear in the total of 6 transactions)
  - Milk: Appears in 4 out of 6 transactions (66.7%).
  - Butter: Appears in 4 out of 6 transactions (66.7%).
  - {Milk, Butter}: Both appear together in 3 out of 6 transactions (50%).
2. Confidence: when x what howmany times y was bought
   1. Confidence measures the strength of the implication. 
   2. It asks: Of the people who bought Milk, what percentage also bought Butter?
   3. Formula:() Support{Milk intersection Butter})/(Support{Milk})
   4. Calculation: 0.5/0.667 = 0.75 => (75%)
   5. Meaning: If a customer has Milk in their cart, there is a 75% chance they will also buy Butter.
3. Lift: Lift tells you if the relationship is actually meaningful or just a coincidence based on how popular Butter is anyway
   1. Formula: $\frac{\text{Confidence}(\text{Milk} \rightarrow \text{Butter})}{\text{Support}(\text{Butter})
   2. Calculation: $\frac{0.75}{0.667} = 1.125
   3. Meaning: Since the Lift is greater than 1, there is a positive association. Buying Milk makes a customer 1.125x more likely to buy Butter than an average customer chosen at random.
4. Key Takeaways:Confidence can be misleading if the second item (Butter) is very popular on its own.Lift is the "sanity check"—if Lift was 1, it would mean Milk and Butter are bought independently, and the "75% confidence" was just because everyone buys Butter regardless.
5. Leverage
6. Conviction: Strength of 
   1. Conviction(X->Y) = (1-Support(Y))/(1-Confidence(X->Y))
   2. I dont want Y to be purchased regardless of X in bulk and hence the numerator is no of times Y was not bpought
   3. we want Y to be bought whenever X was purchased and hence we want the dinominator to be low where we have percentage of times x was bought and y was not bought
