```python
df = pd.read_clipboard()
df
```

```python
X = df.values
```

```python
sklearn.__version__
```

```python
wss = []
for i in range(2,10):
    model = KMeans(n_clusters = i, random_state = 42)
    model.fit(X)
    wss.append(model.inertia_)
```

```python
plt.plot( range(2,10), wss, marker = '*')
plt.xlabel("K value")
plt.ylabel("WSS")
plt.show()
```

```python
from sklearn.metrics import silhouette_score
model = KMeans(n_clusters = 3, random_state = 42)
model.fit(X)
```

```python
silhouette_score(X, model.labels_)
```

```python
from sklearn.metrics import silhouette_score

for i in range(2,10):
    model = KMeans(n_clusters = i, random_state = 42)
    model.fit(X)
    print(f"Silhouette Score for K - {i}",silhouette_score(X, model.labels_))
```
