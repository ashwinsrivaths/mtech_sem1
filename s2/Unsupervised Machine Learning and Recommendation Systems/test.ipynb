```python
# removing outlier in yearly income
df.Yearly_Income.describe()
```

```python

df = df[df.Yearly_Income < 137000].reset_index(drop=True)
# after removing outlier I am resetting the index to maintain a clean DataFrame. 
# This will help in any future operations that rely on the index, such as slicing or merging.
```

```python
# KMeans dosent work well with categorical data, 
# so I am removeing gender column for clustering.

# df = df.drop(columns=['Gender', 'CustomerID'])

df = df[['Yearly_Income','Cust_Spend_Score', 'Age']]
```

```python
# KMeans is sensitive to the scale of the data, so it's important to standardize the features before applying K-Means.
for i in df.select_dtypes(include="number").columns:
    df[i] = df[i].apply(lambda x: (x - df[i].mean())/df[i].std())
```

```python
df.head()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score


wcss = []
for i in range(2, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++',
                    #  max_iter=300, n_init=10, 
                     random_state=42)
    kmeans.fit(df)
    wcss.append(kmeans.inertia_)
    print(f'Number of clusters: {i}, WCSS: {kmeans.inertia_}, Silhouette Score: {silhouette_score(df, kmeans.labels_)}')

wcss

```

```python
plt.plot(range(2, 11), wcss, marker='*')
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
from yellowbrick.cluster import SilhouetteVisualizer

for i in range(2, 11):
    model = KMeans(n_clusters=i, init='k-means++', random_state=42)
    visualizer = SilhouetteVisualizer(model)
    visualizer.fit(df)
    # visualizer.show() # shows everything 
```

```python
kmeand = KMeans(n_clusters=5, init='k-means++', random_state=42)
kmeand.fit(df)
kmeand.inertia_
```

```python
from sklearn.cluster import AgglomerativeClustering

from scipy.cluster.hierarchy import dendrogram, linkage, cophenet


```

```python
df = pd.read_excel("./customer.xlsx").select_dtypes(include="number").drop(columns=['Sex'])

import sklearn

standardscaler = sklearn.preprocessing.StandardScaler()
df = standardscaler.fit_transform(df)
```
